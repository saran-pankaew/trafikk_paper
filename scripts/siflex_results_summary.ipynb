```python
# Directories

    # input
pathway_scores_file = "results/Sanger-2024/06_siflex/output/pathway_scores.csv"
node_scores_file = "results/Sanger-2024/06_siflex/output/node_scores.csv"
node_scores_agg_file = "results/Sanger-2024/06_siflex/output/node_scores_agg.csv"
case_annotations_file = "results/Sanger-2024/06_siflex/output/case_annotations.csv"
    # output|
metadata_output_dir = "tables/"
```

### Pathway scores

```python
pathway_scores = pandas.read_csv(pathway_scores_file)
case_annotations = pandas.read_csv(case_annotations_file)

print("\nPathway Scores DataFrame shape:")
print(pathway_scores.shape)
print("\nUnperturbed min and max values:")
pathway_scores_unperturbed = pathway_scores[pathway_scores['perturbation'] == 'UNPERTURBED']
print(pathway_scores_unperturbed.min())
print(pathway_scores_unperturbed.max())
print("\nPerturbed min and max values:")
pathway_scores_perturbed = pathway_scores[pathway_scores['perturbation'] != 'UNPERTURBED']
print(pathway_scores_perturbed.min())
print(pathway_scores_perturbed.max())

print("\nCase Annotations DataFrame shape:")
print(case_annotations.shape)
print(case_annotations.columns)
```

### Node scores

```python
node_scores = pandas.read_csv(node_scores_file)
print("\nNode Scores DataFrame shape:")
print(node_scores.shape)
print(node_scores.columns)

node_scores_agg = pandas.read_csv(node_scores_agg_file)
print("\nNode Scores Aggregated DataFrame shape:")
print(node_scores_agg.shape)
print(node_scores_agg.columns)
```

#### BCL2i + PI3Ki

```python
cell_lineA = 'NUGC4'
cell_lineB = 'HO1U1'
combo = "PD_01_PD_08"
drugA = "PD_01"
drugB = "PD_08"

# Filter the DataFrame for the specific cell lines, combo, single drugs
node_scores_filtered = node_scores[(node_scores['cell_line'] == cell_lineA) & 
                                    (node_scores['perturbation'] == combo) |
                                    (node_scores['cell_line'] == cell_lineB) &
                                    (node_scores['perturbation'] == combo) |
                                    (node_scores['cell_line'] == cell_lineA) &
                                    (node_scores['perturbation'] == drugA) |
                                    (node_scores['cell_line'] == cell_lineB) &
                                    (node_scores['perturbation'] == drugB) |
                                    (node_scores['cell_line'] == cell_lineA) &
                                    (node_scores['perturbation'] == drugB) |
                                    (node_scores['cell_line'] == cell_lineB) &
                                    (node_scores['perturbation'] == drugA)
                                    ]
node_scores_filtered_T = node_scores_filtered.transpose()
node_scores_filtered_T.to_csv(metadata_output_dir + "node_scores_HO1U1_NUGC4.csv", index=True)
```

#### AKTi + MEKi

```python
drugA = 'PD_06'
drugB = 'PD_10'
combo = "PD_06_PD_10"
combo_r = "PD_10_PD_06"
case_class = "True Positive"

# Filter the node data for the specific combo and single drugs
node_scores_filtered = node_scores[(node_scores['perturbation'] == combo) | (node_scores['perturbation'] == drugA) | (node_scores['perturbation'] == drugB)]
print("\nFiltered Node Scores by combination: {}".format(combo))
print(node_scores_filtered.shape)

# Filter the case annotations for the specific combo and case class
case_annotations_filtered = case_annotations[(case_annotations['perturbation'] == combo_r) &
                                              (case_annotations['classification'] == case_class)]
print("\nFiltered Case Annotations by combination: {} and case class: {}".format(combo_r, case_class))
print(case_annotations_filtered.shape)

# Filter the node scores to include only the cell lines that are in the filtered case annotations
node_scores_filtered = node_scores_filtered[node_scores_filtered['cell_line'].isin(case_annotations_filtered['cell_line'])]
print("\nFiltered Node Scores after filtering by case annotations:")
print(node_scores_filtered.shape)
node_scores_filtered.to_csv(metadata_output_dir + "node_scores_AKTi_MEKi.csv", index=False)

# Node aggregated scores for the specific combo and case class
node_scores_agg_filtered = node_scores_agg[(node_scores_agg['perturbation'] == combo)]
node_scores_agg_filtered = node_scores_agg_filtered[node_scores_agg_filtered['cell_line'].isin(case_annotations_filtered['cell_line'])]
```
